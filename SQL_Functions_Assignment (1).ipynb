## Create Table

```

CREATE TABLE Student_Performance (
    student_id INT PRIMARY KEY,
    name VARCHAR(50),
    course VARCHAR(30),
    score INT,
    attendance INT,
    mentor VARCHAR(50),
    join_date DATE,
    city VARCHAR(50)
);
```

## Q1 Ranking

```
SELECT name, score, RANK() OVER (ORDER BY score DESC) AS rank FROM Student_Performance;
```

## Q2 Previous Score

```
SELECT name, score, LAG(score) OVER (ORDER BY score DESC) AS previous_score FROM Student_Performance;
```

## Q3 Upper + Month

```
SELECT UPPER(name) AS name_upper, MONTHNAME(join_date) AS join_month FROM Student_Performance;
```

## Q4 Next Attendance

```
SELECT name, attendance, LEAD(attendance) OVER (ORDER BY attendance DESC) AS next_attendance FROM Student_Performance;
```

## Q5 NTILE

```
SELECT name, score, NTILE(4) OVER (ORDER BY score DESC) AS performance_group FROM Student_Performance;
```

## Q6 Row Number by Course

```
SELECT name, course, attendance, ROW_NUMBER() OVER (PARTITION BY course ORDER BY attendance DESC) AS row_num FROM Student_Performance;
```

## Q7 Days Enrolled

```
SELECT name, DATEDIFF('2025-01-01', join_date) AS days_enrolled FROM Student_Performance;
```

## Q8 Format Date

```
SELECT name, DATE_FORMAT(join_date, '%M %Y') AS formatted_date FROM Student_Performance;
```

## Q9 Replace City

```
SELECT name, REPLACE(city, 'Mumbai', 'MUM') AS city_updated FROM Student_Performance;
```

## Q10 First Value

```
SELECT name, course, score, FIRST_VALUE(score) OVER (PARTITION BY course ORDER BY score DESC) AS highest_score FROM Student_Performance;
```
